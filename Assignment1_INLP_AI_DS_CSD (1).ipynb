## 28.

◑ Define a function `percent(word, text)` that calculates how often a given word occurs in a text, and expresses the result as a percentage.

````python
# write your function here
def percent(word, text):
    word = word.lower()
    return 100 * sum(1 for w in text if w.lower() == word) / len(text)
````

````python
percent('the', text1)
````

````python
percent('sea', text1)
````

◑ We have been using sets to store vocabularies. Try the following Python expression: `set(sent3) < set(text1)`. Experiment with this using different arguments to `set()`. What does it do? Can you think of a practical application for this?

````python
set(sent3) < set(text1)
````

````python
set(sent3) <= set(text1)   # subset or equal (non-strict version)
set(sent3) > set(text1)    # is sent3's vocab a superset of text1's? (almost certainly False, text1 is huge)
set(sent3) & set(text1)    # intersection — words appearing in BOTH
set(sent3) - set(text1)    # words in sent3 but NOT in text1 (if any)
````
